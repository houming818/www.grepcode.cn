### 1.2 概要

测试目的是在openstack上新建一个nginx的服务，提供html服务。


### 2.1 安装

```shell
$ wget -c https://releases.hashicorp.com/terraform/0.14.10/terraform_0.14.10_linux_amd64.zip
$ unar terraform_0.14.10_linux_amd64.zip
$ sudo cp terraform /usr/bin/

# 验证是否安装成功
$ terraform --version
Terraform v0.14.10
```


### 2.2 初始化项目目录

```shell
$ mkdir hello_terra
$ cd hello_terra
$ vim provider.tf

# ---- provider.tf ---- 模板
terraform {
  required_version = ">= 0.14.0"
  required_providers {
    openstack = {
      source  = "terraform-provider-openstack/openstack"
      version = "~> 1.35.0"
    }
  }
}

provider "openstack" {
  user_name   = "admin"
  tenant_name  = "admin"
  password    = "********"
  auth_url    = "http://{{HOST_IP}}/v2.0"
  region     = "RegionOne"
}
# -----

$ terraform init

Initializing the backend...

Initializing provider plugins...
- Finding terraform-provider-openstack/openstack versions matching "~> 1.35.0"...
- Installing terraform-provider-openstack/openstack v1.35.0...
- Installed terraform-provider-openstack/openstack v1.35.0 (self-signed, key ID 4F80****)

Partner and community providers are signed by their developers.
If you'd like to know more about provider signing, you can read about it here:
https://www.terraform.io/docs/cli/plugins/signing.html

Terraform has created a lock file .terraform.lock.hcl to record the provider
selections it made above. Include this file in your version control repository
so that Terraform can guarantee to make the same selections by default when
you run "terraform init" in the future.

Terraform has been successfully initialized!

You may now begin working with Terraform. Try running "terraform plan" to see
any changes that are required for your infrastructure. All Terraform commands
should now work.

If you ever set or change modules or backend configuration for Terraform,
rerun this command to reinitialize your working directory. If you forget, other
commands will detect it and remind you to do so if necessary.

```
