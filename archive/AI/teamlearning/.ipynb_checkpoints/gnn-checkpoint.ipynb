## 话题一 图论与程序中图的操作

### 第二章 简单图论

#### 数据结构的扩充

由深度学习的原有结构：**矩阵、张量、序列（sequence）、间序列（time series）** 扩充新的 **图** 结构

#### 解决的基本问题

**节点分类问题 边预测问题**

与传统机器学习的差别：

图节点样本是非独立或非相同分布的，因此传统机器学习方法不适用于图计算任务。

解决思路：

**构建节点表征**，节点表征要求同时包含节点自身的信息和节点邻接的信息。图机器学习成功的关键在于如何为节点构建表征。利用神经网络来学习节点表征。

#### 图的数据结构

节点和边的信息： 类别型（categorical、标签label） 或者 数值型（numeric，属性attribute）

1.**定义一** 图 

![定义图](../../images/teamlearning_gnn_01.png)

2.**定义二** 邻接矩阵

- 有向图 无向图
- 有权图 无权图

#### 图的属性

1.**定义三** 节点的度，degree

2.**定义四** 邻接节点，neighbors

3.**定义五** 路径，行走，walk

4.**定理六** 没看懂

```
有一图，其邻接矩阵为 $\mathbf{A}$, $\mathbf{A}^{n}$为邻接矩阵的$n$次方，那么$\mathbf{A}^{n}[i,j]$等于从节点$v_i$到节点$v_j$的长度为$n$的行走的个数。（也就是，以节点$v_i$为起点，节点$v_j$为终点，长度为$n$的节点访问方案的数量，节点访问中可以兜圈子重复访问一些节点）
```
5.**定义七** 路径，path

6.**定义八** 子图，subgraph

7.**定义九** 连通分量，connected component

8.**定义十** 连通图，connected graph

9.**定义十一** 最短路径，shortest path

10.**定义十二** 直径，diameter

11.**定义十三** 拉普拉斯矩阵，Laplacian Matrix 没看懂

12.**定义十四** 对称归一化的拉普拉斯矩阵，Symmetric normalized Laplacian

#### 图的种类

- 同质图（Homogeneous Graph）：只有一种类型的节点和一种类型的边的图
- 异质图（Heterogeneous Graph）：存在多种类型的节点和多种类型的边的图
- 二部图（Bipartite Graphs）：节点分为两类，只有不同类的节点之间存在边。

#### 图的算法

机器学习

1. 节点预测：预测节点的类别或某类属性的取值
  1. 例子：对是否是潜在客户分类、对游戏玩家的消费能力做预测
2. 边预测：预测两个节点间是否存在链接
  1. 例子：Knowledge graph completion、好友推荐、商品推荐
3. 图的预测：对不同的图进行分类或预测图的属性
  1. 例子：分子属性预测
4. 节点聚类：检测节点是否形成一个社区
  1. 例子：社交圈检测
5. 其他任务
  1. 图生成：例如药物发现
  2. 图演变：例如物理模拟
  3. ……
  
#### 应用神经网络于图面临的挑战

> 图数据是非规则的非结构化的
>
> 1. 任意的大小和复杂的拓扑结构；
> 2. 没有固定的节点排序或参考点；
> 3. 通常是动态的，并具有多模态的特征；
> 4. 图的信息并非只蕴含在节点信息和边的信息中，图的信息还包括了图的拓扑结构。

> 在此篇文章中，我们学习了简单的图论知识。对于学习此次组队学习后续的内容，掌握这些图论知识已经足够。如果有小伙伴希望掌握更多的图
