```python
import numpy as np

# 3. 从所研究总体中采样一个随机样本
samples = np.random.normal(0, 1, size=30)
```

```python
# 4. 在承认原假设的前提下，构造检验样本统计量
from scipy.stats import t
def calc_tau(data, u=0):
    mean_val = np.mean(data)
    se = np.std(data)/np.sqrt(len(data))
    df = len(data) -1  # 自由度
    p_val = (mean_val-mu)/se
    tau = t.cdf(p_val, df)
    return tau
```

```python
# 5. 输入拒绝域临界值
# 6. 输出结果

mu = 0
refuse = 0.05
tau = calc_tau(samples, mu)

tau > refuse
```
