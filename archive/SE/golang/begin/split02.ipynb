> 前四章为一分，主要是说明组成Golang的砖。第二分将说明如何将这些砖组成为逻辑代码。

不太理解闭包的优势

```go
import "fmt"

type Apple struct {
    Weight int
}

func AddOnCopy(a Apple) (Apple) {
    a.Weight += 1
    fmt.Printf("change to %v\n", a)
    return a
}

func AddOnAddress(a *Apple) (*Apple) {
    a.Weight += 1
    fmt.Printf("change to %v\n", a)
    return a
}

func test01(){
    var apple Apple
    fmt.Printf("Before AddOnCopy %v\n", apple)
    AddOnCopy(apple)
    fmt.Printf("After AddOnCopy %v\n", apple)
    
    fmt.Printf("Before AddOnAddress %v\n", apple)
    AddOnAddress(&apple)
    fmt.Printf("After AddOnAddress %v\n", apple)
}

test01()
```

#### 第七章 接口(interface)

```go
import "fmt"

type DataWriter interface {
    WriteData(data interface{}) error
    FlushData() error
}

type file struct {
   DataWriter
}

// 实现DataWriter接口的WriteData()方法
func (d *file) WriteData(data interface{}) error {
    fmt.Println("WriteData: ", data)
    return nil
}

func test02(){
    // 实例化 file
    f := new(file)
    
    var writer DataWriter
    
    writer = f
    
    writer.WriteData("data")
}

test02()

fmt.Println(`
repl.go:24:5: error compiling assignment: writer = f
	repl.go:24:5: incompatible types in assignment: main.DataWriter = *main.file
	reason: *main.file does not implement main.DataWriter: missing method func (interface{FlushData() error; WriteData(interface{}) error}).FlushData() error
`)
```
